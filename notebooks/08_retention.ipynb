```python
import pandas as pd
from pathlib import Path

project = Path.cwd().parent

predictions = pd.read_csv(
    project / "outputs" / "churn_predictions.csv"
)

high_risk = predictions[
    predictions["CHURN_PROBABILITY"] >= 0.70
].copy()

high_risk["RISK_AFTER_INTERVENTION"] = (
    high_risk["CHURN_PROBABILITY"] * 0.80
).clip(0, 1)

simulation = pd.DataFrame({
    "Metric": [
        "High-risk customers",
        "Average churn probability before",
        "Average churn probability after",
        "Average risk reduction",
        "Estimated customers moved below 0.70 risk"
    ],
    "Value": [
        len(high_risk),
        round(high_risk["CHURN_PROBABILITY"].mean(), 4),
        round(high_risk["RISK_AFTER_INTERVENTION"].mean(), 4),
        round(
            (
                high_risk["CHURN_PROBABILITY"].mean()
                - high_risk["RISK_AFTER_INTERVENTION"].mean()
            ),
            4
        ),
        (
            high_risk["RISK_AFTER_INTERVENTION"] < 0.70
        ).sum()
    ]
})

high_risk.to_csv(
    project / "outputs" / "retention_simulation_customers.csv",
    index=False
)

simulation
```
